```python
#crear nueva funcion biologica
def ContenidoGC(DNA):
  counter = 0
  if DNA[0] =='C' or DNA[0] == 'G':
    counter = counter + 1
  if DNA[1] =='C' or DNA[1] == 'G':
    counter = counter + 1
  if DNA[2] =='C' or DNA[2] == 'G':
    counter = counter + 1
  if DNA[3] =='C' or DNA[3] == 'G':
    counter = counter + 1
    return counter / 4.0
```
